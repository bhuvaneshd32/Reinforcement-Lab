```python
import numpy as np
```

```python
num_docs = 100
k = 10
epsilon = 0.2
delta = 0.7
x=1000
prob = np.random.rand(k,num_docs)
```

```python
def algo(documents,k,x,prob):
  documents_visted = []
  rank = []

  for i in range(k):
    counts = np.zeros(len(documents))
    for j in range(len(documents)):
      if j in documents_visted:
        continue
      p = prob[i][j]
      counts[j] = np.random.binomial(n=x,p=p)
    doc_index = np.argmax(counts)
    doc = documents[doc_index]
    rank.append(doc)
    documents_visted.append(doc)
    print(f"Rank {i} , Document Selected {doc}")
  return rank


num_docs = 100
k = 10
eps = 0.2
delta = 0.7
x=((2*k*k)/(eps*eps))*np.log((2*k)/delta) #sample formula given in question
prob = np.random.rand(k,num_docs)
documents = []
for i in range(num_docs):
  documents.append(i)

algo(documents,k,x,prob)
```
