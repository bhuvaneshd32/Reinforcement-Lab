```python
import numpy as np
```

```python
# Q - Array to store averages
# N - Array to store counts
# t - the current time step

means = [0.1, 0.3, 0.9, 0.7, 0.5] # True means of the arms
k = len(means) # Number of arms
num_pulls = 1000
```

```python
def get_reward(arm):
  return np.random.normal(means[arm],1.0)
```

```python
# -- 1. Incremental Uniform (Round Robin)---
def incremental_uniform():

  Q = np.zeros(k)
  N = np.zeros(k)

  for t in range(num_pulls):

    arm = t%k
    reward = get_reward(arm)
    N[arm] += 1
    Q[arm] += (reward-Q[arm])/N[arm]

  return np.argmax(Q)

```

```python
# --- 2. Epsilon Greedy ---
def epsilon_greedy(epsilon=0.1):

  Q = np.zeros(k)
  N = np.zeros(k)

  for t in range(num_pulls):

    if np.random.random() < epsilon:
      arm = np.random.randint(k) # Explore
    else:
      arm = np.argmax(Q) # Exploit

    reward = get_reward(arm)
    N[arm] += 1
    Q[arm] += (reward- Q[arm])/ N[arm]

  return np.argmax(Q)
```

```python
# --- 3. UCB Algorithm ---
def ucb(c=1):

  N = np.zeros(k)
  Q = np.zeros(k)

  for arm in range(0,k):
    N[arm]+=1
    Q[arm] = get_reward(arm)

  for t in range(k,num_pulls):
    ucb_values = Q + c*np.sqrt(np.log(t)/N)
    arm = np.argmax(ucb_values)

    reward = get_reward(arm)
    N[arm] += 1
    Q[arm] += (reward - Q[arm]) / N[arm]

  return np.argmax(Q)

```

```python
print(f"Uniform Best Arm: {incremental_uniform()}")
print(f"E-Greedy Best Arm: {epsilon_greedy()}")
print(f"UCB Best Arm: {ucb()}")
```
